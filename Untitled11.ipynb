```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'])
monthly_revenue = df.groupby(df['Order_Date'].dt.to_period('Y'))['Revenue_USD'].sum()
monthly_revenue
```

```python
monthly_revenue
```

```python
#convert the index
monthly_revenue.index=(
    monthly_revenue.index.astype(str)

)
```

```python
monthly_revenue
```

```python
from matplotlib.lines import lineMarkers
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))
plt.plot(
    monthly_revenue.index,
    monthly_revenue.values,
    marker='*',
    color='red',
    linestyle='--',
    linewidth=2,
    markersize=10
)
plt.title("year Revenue",fontsize=30)
plt.xlabel("year",fontsize=30)
plt.ylabel("Revenue")
plt.grid(True, alpha=0.4)
plt.show()
```

```python
# WHICH PRODUCT CATEGORY  GENERATE MOST REVENUE
category_revenue=(
    df.groupby('Category')['Revenue_USD']
    .sum()
     .sort_values(ascending=False)
)
category_revenue
```

```python
from matplotlib.typing import ColourType
plt.figure(figsize=(10,5))

plt.bar(
    category_revenue.index,
    category_revenue.values



)
plt.title("Category  by  Revenue",fontsize=30)
plt.xlabel("Category",fontsize=15)
plt.ylabel("Revenue",fontsize=15)
plt.grid(True, alpha=0.4)
plt.show()
```

```python
#horizontal bar chart
from matplotlib.typing import ColourType
plt.figure(figsize=(10,6))

plt.barh(
    category_revenue.index,
    category_revenue.values



)
plt.title("Category  by  Revenue",fontsize=30)
plt.xlabel("Category",fontsize=15)
plt.ylabel("Revenue",fontsize=15)
plt.grid(True, alpha=0.4)
plt.show()
```
