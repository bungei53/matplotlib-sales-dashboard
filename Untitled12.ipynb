```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df= pd.read_csv('matplotlib_practice_dataset_3000x22 (1).csv')
df.head()
```

```python
df.columns
```

```python
category_revenue=(
    df.groupby('Category')['Revenue_USD'].sum().sort_values(ascending= False)
)
```

```python
plt.figure(figsize=(10,5))
plt.plot(
    category_revenue.index,
    category_revenue.values,
    marker='*',
    color='#533A7B',
    linestyle='-.',
    linewidth=2
)
plt.title("category VS revenue",fontsize=15)
plt.xlabel("category", fontsize=15)
plt.ylabel('revenue', fontsize=15)
plt.grid(True,alpha=0.5)
plt.show()
```

```python
category_count=df['Category'].value_counts()
category_count
```

```python
custom_colors = ['#8195C6', '#757FBD', '#6F74B8', '#6C6FB6', '#6969B3']
plt.figure(figsize=(8,8))
plt.pie(
    category_count.values,
    labels=category_count.index,
    autopct='%1.1f%%',
    startangle=90,
    colors=custom_colors  # Pass the custom colors list here
)
plt.title('PRODUCT DISTRIBUTION BY CATEGORY',fontsize=15)
plt.show()
```

```python
# Create a figure with 1 row and 2 columns for side-by-side subplots
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(18, 8))

# --- Plot 1: Line Chart of Category vs Revenue ---
axes[0].plot(
    category_revenue.index,
    category_revenue.values,
    marker='*',
    color='#533A7B',
    linestyle='-.',
    linewidth=2
)
axes[0].set_title("Category vs Revenue", fontsize=14, fontweight='bold')
axes[0].set_xlabel("Category", fontsize=12)
axes[0].set_ylabel("Revenue (USD)", fontsize=12)
axes[0].grid(True, alpha=0.4)

# --- Plot 2: Pie Chart of Product Distribution ---
custom_colors = ['#8195C6', '#757FBD', '#6F74B8', '#6C6FB6', '#6969B3']
axes[1].pie(
    category_count.values,
    labels=category_count.index,
    autopct='%1.1f%%',
    startangle=90,
    colors=custom_colors
)
axes[1].set_title("Product Distribution by Category", fontsize=14, fontweight='bold')

# Adjust layout to fit both charts neatly
plt.tight_layout()
plt.show()
```

```python
import numpy as np

fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(16, 12))

# --- 1. Top-Left (Row 0, Col 0): Category vs Revenue (Line Plot) ---
axes[0, 0].plot(
    category_revenue.index,
    category_revenue.values,
    marker='*',
    color='#533A7B',
    linestyle='-.',
    linewidth=2
)
axes[0, 0].set_title("Category vs Revenue", fontsize=12, fontweight='bold')
axes[0, 0].set_xlabel("Category")
axes[0, 0].set_ylabel("Revenue (USD)")
axes[0, 0].grid(True, alpha=0.3)

# --- 2. Top-Right (Row 0, Col 1): Product Distribution (Pie Chart) ---
custom_colors = ['#8195C6', '#757FBD', '#6F74B8', '#6C6FB6', '#6969B3']
axes[0, 1].pie(
    category_count.values,
    labels=category_count.index,
    autopct='%1.1f%%',
    startangle=90,
    colors=custom_colors
)
axes[0, 1].set_title("Product Distribution by Category", fontsize=12, fontweight='bold')

# --- 3. Bottom-Left (Row 1, Col 0): Sales Channel Distribution (Bar Chart) ---
channel_counts = df['Sales_Channel'].value_counts()
axes[1, 0].bar(channel_counts.index, channel_counts.values, color='#757FBD', alpha=0.8)
axes[1, 0].set_title("Sales Channel Distribution", fontsize=12, fontweight='bold')
axes[1, 0].set_xlabel("Channel")
axes[1, 0].set_ylabel("Count")
axes[1, 0].grid(axis='y', alpha=0.3)

# --- 4. Bottom-Right (Row 1, Col 1): Unit Price vs Profit (Scatter Plot with Abline) ---
axes[1, 1].scatter(df['Unit_Price_USD'], df['Profit_USD'], color='#533A7B', alpha=0.5, edgecolor='none', label='Data Points')

# Calculate the linear regression line (y = mx + b)
x = df['Unit_Price_USD']
y = df['Profit_USD']
m, b = np.polyfit(x, y, 1)

# Plot the trend line
axes[1, 1].plot(x, m*x + b, color='#D14949', linestyle='--', linewidth=2.5, label=f'Trend: y={m:.2f}x+{b:.2f}')

axes[1, 1].set_title("Unit Price vs Profit", fontsize=12, fontweight='bold')
axes[1, 1].set_xlabel("Unit Price (USD)")
axes[1, 1].set_ylabel("Profit (USD)")
axes[1, 1].grid(True, alpha=0.3)
axes[1, 1].legend()

plt.tight_layout()
plt.show()
```
