```python
# Reload the data fresh to undo any half-applied changes
df = pd.read_csv('kenya_ecommerce_dirty_4000x30.csv')

# Convert dates properly
df['Order_Date'] = pd.to_datetime(df['Order_Date'], errors='coerce')
df['Ship_Date'] = pd.to_datetime(df['Ship_Date'], errors='coerce')

# Handle missing values in numerical columns by imputing the median
num_cols_with_na = ['Customer_Age', 'Delivery_Days', 'Customer_Rating', 'Marketing_Cost_KES', 'Inventory_At_Order']
for col in num_cols_with_na:
    df[col] = df[col].fillna(df[col].median())

# Handle missing values in categorical columns by filling with 'Unknown'
cat_cols_with_na = ['Gender', 'City', 'Campaign']
for col in cat_cols_with_na:
    df[col] = df[col].fillna('Unknown')

# Check for any remaining missing values
display(df.isnull().sum())
```

```python
df.columns
```

```python
# 1. Fill Unit_Price_KES based on Product_ID median price
df['Unit_Price_KES'] = df['Unit_Price_KES'].astype(str).str.replace(',', '').str.extract(r'([\d\.]+)').astype(float)
df['Unit_Price_KES'] = df.groupby('Product_ID')['Unit_Price_KES'].transform(lambda x: x.fillna(x.median()))
df['Unit_Price_KES'] = df['Unit_Price_KES'].fillna(df['Unit_Price_KES'].median())

# 2. Handle missing Order_Date by chronological forward-fill (or interpolation)
df['Order_Date'] = df['Order_Date'].ffill().bfill()

# 3. Handle missing Ship_Date by adding Delivery_Days to Order_Date
missing_ship_mask = df['Ship_Date'].isnull()
pending_delivery_days = pd.to_timedelta(df['Delivery_Days'], unit='D')
df.loc[missing_ship_mask, 'Ship_Date'] = df.loc[missing_ship_mask, 'Order_Date'] + pending_delivery_days.loc[missing_ship_mask]

# Check missing values again
display(df.isnull().sum())
```

```python

df['Category'] = df['Category'].str.title()


category_total = df.groupby('Category')['Total_Amount_KES'].sum()


plt.figure(figsize=(10, 6))
plt.pie(
    category_total.values,
    labels=category_total.index,
    autopct='%1.1f%%',
    startangle=140
)
plt.title("Total Amount KES by Category")
plt.tight_layout()
plt.show()
```

```python

df['Category'] = df['Category'].str.title()
category_total = df.groupby('Category')['Total_Amount_KES'].sum()

fig, ax = plt.subplots(2, 2, figsize=(12, 10))


ax[0, 0].scatter(
    df['Customer_Age'],
    df['Total_Amount_KES'],
    color='pink',
    alpha=0.5
)
ax[0, 0].set_title('Customer Age vs Total Amount (KES)')
ax[0, 0].set_xlabel('Age')
ax[0, 0].set_ylabel('Total Amount')
ax[0, 0].grid(True,alpha=0.5)

# Line plot of Category Totals
ax[0, 1].plot(
    category_total.index,
    category_total.values,
    color='red',
    linewidth=4
)
ax[0, 1].set_title('Category Totals Trend')
ax[0, 1].set_xlabel('Category')
ax[0, 1].set_ylabel('Total Amount (KES)')
ax[0, 1].tick_params(axis='x', rotation=45),
ax[0, 1].grid(True,alpha=0.2)

# Bar plot of Category Totals
ax[1, 0].bar(
    category_total.index,
    category_total.values,
    color='green'
)
ax[1, 0].set_title('Category Totals Bar Chart')
ax[1, 0].set_xlabel('Category')
ax[1, 0].set_ylabel('Total Amount (KES)')
ax[1, 0].tick_params(axis='x', rotation=45),
ax[1, 0].grid(True,alpha=0.3)

# Histogram of Customer Age distribution
ax[1, 1].hist(
    df['Customer_Age'].dropna(),
    bins=15,
    color='purple',
    edgecolor='black'
)
ax[1, 1].set_title('Customer Age Distribution',fontweight='bold',fontsize=10)
ax[1, 1].set_xlabel('Age')
ax[1, 1].set_ylabel('Frequency'),
ax[1, 1].grid(True,alpha=0.4)

plt.tight_layout()
plt.show()
```
